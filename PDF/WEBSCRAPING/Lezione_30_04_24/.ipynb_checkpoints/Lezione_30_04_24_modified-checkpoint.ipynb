## PANDAS: memory model

* il modello di memoria di Pandas ha origine da quello di numpy, ma verrà modificato a partire dalla versione **3.0**  e dalle versione **2.0** sarà facilmente settabile
* numpy
    * quando si crea un nuovo array a partire da uno esistente, si crea **una vista** dell'array originario
    * la vista fa riferimento all'array originario ed ogni modifica all'array originario o alla vista creata si ripercuote sull'altro array
meaning changes to either the original array or the view have the potential to impact one another.
* in PANDAS, **Series** e **DataFrame** sono **backed** da array numpy, il comportamento è simile


```python
import pandas as pd
print(pd.__version__)
```

## PANDAS: memory model

* senza la opzione **CopyOnWrite** il comportamneto di **Pandas** è molto difficile da prevedere
    * alcune operazioni **Pandas** restituiscono una **copia** della struttura dati, altre una **view**
* se un **oggetto A** è la copia di un **oggetto B** allora ognuno ha una divresa parte di memoria allocat
    * quando cambio modifico la copia non modifico il dato originario
* se un **oggetto A** è una **view** dell'oggetto **B**, allora condividono la stessa zono di memoria, ed ogni modifica ad 
**B** modifica anche l'oggetto originario
* importante distinguere perchè
    * effetti laterali nel codice
    * **efficienza**: se si lavora con **DataSet** grandi, è possibile che non si voglia generare una copia, a causa della occupazione di memoria
* molto difficile prevedere quando **Pandas** effettua una copia e quando invece **una view**




## DataFrame: copy or view?

```python
import pandas as pd

# This is the default option so I don't strictly
# need this command, but I'll add it to be explicit
pd.set_option("mode.copy_on_write", False)

df = pd.DataFrame({"a": [10, 20, 30, 40], "b": [11, 12, 13, 14]})
df
```

```python
my_slice = df.iloc[1:3,]
my_slice
```

## DataFrame: copy or view?

```python
df.iloc[1, 1] = -1
df
```

```python
my_slice
```

* l'esecuzione del programma ha prodotto una view sul DataFrame originario
* i due **DataFrame** condividono lo stesso spazio di memoria


## DataFrame: copy or view?

```python
# Same DataFrame and subset:
df = pd.DataFrame({"a": [10, 20, 30, 40], "b": [11, 12, 13, 14]})
my_slice = df.iloc[1:3,]
my_slice
```

```python
# But now we set the value to 3.14 instead of -1.
df.iloc[1, 1] = 3.14
df
```

* questo comportamento dipende dal fatto di poter o meno sostituire il nuovo valore direttamente nella vecchia poszione dell'array
* comportamento difficile da prevedere!


```python
my_slice
```

## DataFrame: copy or view?

```python
df = pd.DataFrame({"a": [10, 20, 30, 40], "b": [11, 12, 13, 14]})
my_slice = df.iloc[1:3,].copy()
my_slice
```

```python
df.iloc[1, 1] = -1
df
```

```python
my_slice
```

* posso **copiare esplicitamente** il DatFrame, tuttavia questo comporterà un notevole aumento della memoria utilizzata


## DataFrame: Copy on Write (CoW)

* le **views** sono un meccanismo che consentono di migliorare l'uso della memoria
* i **DataFrame** sono oggetti di dimensione elvata, quindi fare delle copie costa molto, in termini di **memory consumption**
* le regole di **Panda** non sono chiare, ed il comportamento del programma può risultare non predicibile
* **Copy on Write**
    * cerca di ottenere allo stesso tempo un buon **memory usage** e una **semantica chiara**
* dal punto di vista dell'utente
    * Pandas si comporta cone tutte le operazioni su un **DataFrame** restituissero una **copia**
* dal punto di vista della implementazione
    * **Pandas** non genera **effettivamente** una copia 
    * si crea una **view** e solo se viene apportata una modifica ad uno dei due DataFrame, viene allora effettata una **copia**
    * **views** e **copues** hanno lo stesso effetto fino a che gli array non vengono modificati
    

## DataFrame: Copy on Write (CoW)

```python
import pandas as pd

pd.set_option("mode.copy_on_write", True)
df = pd.DataFrame({"a": [10, 20, 30, 40], "b": [11, 12, 13, 14]})
df
```

```python
my_slice = df.iloc[1:3,]
my_slice
```

## DataFrame: Copy on Write (CoW)

```python
df.iloc[1, 1] = -1
df
```

```python
my_slice
```

## DataFrame: Copy on Write (CoW)

```python
# But now we set the value to 3.14 instead of -1.
df.iloc[1, 1] = 3.14
df
```

```python
my_slice
```

## Ridurre l'uso della RAM in Pandas: scegliere i tipi

* primo, semplice,  accorgimento: utilizzare tipi che richiedano poca memoria
* valori numerici
    * **Pandas** per default memorizza i valori interi come **int64** e i valori decimali come **float64**
* **Downcasting**
    * memorizzare **int64** come **int8**
    * memorizzare **float64** come **float8**
* utilizzare la funzione **astype**    

## Ridurre l'uso di memoria in Pandas: valori numerici

```python
listint=[]
for i in range(100000000): 
    listint.append(i)
seriesint = pd.Series(listint)
print(f"x1 dtype: {seriesint.dtype}")
print(f"x1 memory usage: {seriesint.memory_usage(deep=True)} bytes")
print("")

seriesintopt = pd.Series(listint).astype('int8')
print(f"x1 dtype: {seriesintopt.dtype}")
print(f"x1 memory usage: {seriesintopt.memory_usage(deep=True)} bytes")
print("")
```

* notare l'uso della funzione **memory_usage**

```python
The info() method in Pandas tells us how much memory is being taken up by a particular dataframe. To do this, we can assign the memory_usage argument a value = “deep” within the info() method. This will give us the total memory being taken up by the pandas dataframe.

However, the info() method does not give us a detailed description of the memory usage. It only tells us the total memory being used by the dataframe. For a more detailed overview, we can use the memory_usage() method. The memory_usage() method gives us the total memory being used by each column in the dataframe. It returns a Pandas series which lists the space being taken up by each column in bytes. Passing the deep argument a value = True within the memory_usage() method gives us the total memory usage of the dataframe columns.
```

## Ridurre l'uso di memoria in Pandas: categorical types

```python
import pandas as pd

#Sample DataFrame creation
df =pd.DataFrame({
    'A':range(1,10001),
    'B': [float(i) for i in range (1, 10001)],
    'C':['category'+str(i%10) for i in range(1,10001)]
})
print(df.memory_usage(deep=True))
#before optimization
print("Memory usage before optimization:")
print(df.memory_usage(deep=True).sum())

#Converting to more efficient data types
df['A']=df['A'].astype('int32') #smaller integer type
df['B']=df['B'].astype('float32') #smaller float type
df['C']=df['C'].astype('category') # Converting to category

#After optimization
print("\nMemory usage after optimization:")
print(df.memory_usage(deep=True).sum())
```
